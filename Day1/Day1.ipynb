```python
print("Hello World")

print(2 + 3) #addition(+)
print(3 - 1) #subtraction(-)
print(2 * 3) #multiplication(*)
print(3 + 2) #addition(+)
print(3 - 2) #subtraction(-)
print(3 * 2) #multiplication(*)
print(3 / 2) #division(/)
print(3 ** 2) #expontial(**)
print(3 % 2) #modulus(%)
print(3 // 2) #Floor division operator(//)

#Checking data types

print(type(10)) #Int
print(type(3.14)) #Float
print(type(1 + 3j)) #complex
print(type('Kai')) #String
print(type([1,2,3])) #List
print(type({'name':'Kai'})) #Dictionary
print(type({8.8, 3.14, 2.2})) #Tuple
```
